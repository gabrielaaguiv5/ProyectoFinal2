```python
# =========================================================
# SmartChat Insight
# Análisis de los datos, clientes (recurrentes, perdidos) y productos
# =========================================================

# Importar librerías
import re
import pandas as pd
from pathlib import Path
import matplotlib.pyplot as plt

# Definir rutas
clean_path = Path("../data/cleaned/whatsapp_clean.csv")
output_path = Path("../data/outputs/")
output_path.mkdir(parents=True, exist_ok=True)

# Cargar datos limpios
df = pd.read_csv(clean_path)
df["date"] = pd.to_datetime(df["date"], errors="coerce")

print("Datos cargados correctamente")
print(f"Total de mensajes: {len(df)}")
df.head(5)
```

```python
# Pasar todo a minúsculas
df["clean_message"] = df["message"].str.lower()

# Eliminar puntuación básica
df["clean_message"] = df["clean_message"].apply(lambda x: re.sub(r"[^a-záéíóúüñ0-9\s]", "", x))
```

```python
def remove_emojis(text):
    return re.sub(r"[^\w\s.,!?áéíóúüñÁÉÍÓÚÜÑ]", "", str(text))
```

```python
df["message"] = df["message"].apply(remove_emojis)
```

```python
from collections import Counter
import nltk
from nltk.corpus import stopwords


nltk.download("stopwords")
stopwords_es = set(stopwords.words("spanish"))

# Dividir en palabras y eliminar stopwords
all_words = [
    word for msg in df["clean_message"].dropna()
    for word in msg.split()
    if word not in stopwords_es and len(word) > 2
]

# Contar palabras más comunes
word_counts = Counter(all_words)
pd.DataFrame(word_counts.most_common(30), columns=["word", "count"])
```
